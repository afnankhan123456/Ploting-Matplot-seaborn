# Step chart

```python
import matplotlib.pyplot as plt
```

```python
x=[1, 2, 3, 4, 5, 6]
y=[1, 6, 3, 4 , 5, 2]

plt.step(x,y)
plt.show()
```

```python
# plt.grid() --- is comond se grid bnn jayega

x=[1, 2, 3, 4, 5, 6]
y=[1, 6, 3, 4 , 5, 2]

plt.title("python")
plt.grid()  
plt.step(x,y)
plt.show()
```

```python
# marker='o'  is code se hum o ya * lga sakte he line pr
# mfc='r'   colors ke liye
#  ms=30  dot ka size bda krne ke liye

x=[1, 2, 3, 4, 5, 6]
y=[1, 6, 3, 4 , 5, 2]

plt.title("python")
plt.xlabel("x- axis")
plt.ylabel("y- axis")
plt.grid()  
plt.step(x,y,  marker='*',mfc='r', ms=30 )
plt.show()
```
