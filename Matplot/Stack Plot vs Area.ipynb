# Stack Plot vs Area

```python
import matplotlib.pyplot as plt
```

```python
x= [1, 2, 3, 4, 5, 6, 7]
area = [2, 1, 3, 2, 5, 1, 2]

plt.stackplot(x, area)
plt.show()
```

```python
x= [1, 2, 3, 4, 5, 6, 7]
area1 = [2, 1, 3, 2, 5, 1, 2]
area2 = [2, 1, 3, 2, 5, 1, 2]
area3 = [2, 1, 4, 2, 7, 1, 9]


plt.stackplot(x, area1, area2, area3)
plt.show()
```

```python


x = [1, 2, 3, 4, 5, 6, 7]
area1 = [2, 1, 3, 2, 5, 1, 2]
area2 = [2, 1, 3, 2, 5, 1, 2]
area3 = [2, 1, 4, 2, 7, 1, 9]

plt.title('python')
plt.stackplot(x, area1, area2, area3, labels=['area1', 'area2', 'area3'], colors=['r', 'k', 'b'])  
plt.legend(loc=2)
plt.show()
```
