# Pai Chart

```python
import matplotlib.pyplot as plt
```

```python
# labels=y ---- is comond se side me name aa jayega
# ex=[0.1, 0.0, 0.0, 0.0]  explode=ex --- graph bhar nikl jayege
# autopct='%0.1f%%'  ---- is comond se number ke bad percentage lg jayega
# shadow=True  --- is comond se halki si shadow aa jayegi
# radius=1.5 --- is comond se graph ka size bda or gatha sakta he
# startangle=180  ---- is comond se graph ko rotate kr sakte he

x=[10, 20, 30, 40]
y=['c', 'c++', 'java', 'python']

ex=[0.1, 0.0, 0.0, 0.0]
c=['r', 'b', 'c', 'k']

plt.pie(x, labels=y, explode=ex, colors=c, autopct='%0.1f%%',shadow=True, radius=1.5,startangle=180)
plt.show()
```

```python
x=[30, 40, 50, 70, 90]
y=['c', 'c++', 'java', 'python', 'computer' ]

plt.pie(x)
plt.show()
```

```python
# textprops={'fontsize':15   ---- is comond se hum label ke fontsize ko bda kr sakte he
# plt.title('Wscube Tech')   ---  is comon se tital aa jayega
# plt.legend(loc=2)   ----  is comond se hum legend set kr sakte he

x = [30, 40, 50, 70, 90]
y = ['c', 'c++', 'java', 'python', 'computer']

c = ['g', 'm', 'c', 'y', 'k']  

plt.pie(x, labels=y, colors=c, textprops={'fontsize':15})
plt.title('Wscube Tech')
plt.legend(loc=2)
plt.show()
```

```python
#pie chart ke ander ek or chart bnaye

x=[30, 40, 50, 70, 90]
x2=[15, 20, 25, 35, 50]
y=['c', 'c++', 'java', 'python', 'computer' ]

c=['r', 'g', 'b', 'y']
c2=['r', 'y', 'k', 'm']

plt.pie(x, labels=y, radius=1.5)
plt.pie(x2, colors=c2)
plt.show()
```

# Stem Plots

```python
import matplotlib.pyplot as plt
```

```python
x=[2, 3, 4, 5, 6, 7]
y=[1, 2, 4, 6, 5, 4]

plt.stem(x,y)
plt.show()
```

```python
# linefmt=':'   --- is comond se line dot me badal jayegi
# markerfmt='k*'  --- is comond se hum * ya * or unme color bhi kr sakte he
# basefmt='r'   ---- is comond se hum under line ko color kr sakte he 
# label='Python')
#  plt.legend()    is comond se hum legend lga sakte he
#  orientation='horizontal --- is comond se horizontal ho jayega


x=[2, 3, 4, 5, 6, 7]
y=[1, 2, 4, 6, 5, 4]

plt.stem(x,y, linefmt=':', markerfmt='k*', basefmt='r', label='Python', orientation='horizontal')
plt.legend()
plt.show()
```
