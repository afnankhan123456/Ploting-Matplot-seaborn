# Line Plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

var = [1, 2, 3, 4, 5, 6, 7]
var2 = [2, 3, 4, 1, 5, 6, 7]


df = pd.DataFrame({"var": var, "var2": var2})


sns.lineplot(x='var', y='var2', data=df)

plt.show()

#yha humne dataframe bna kr data ko vusual kiya he
```
