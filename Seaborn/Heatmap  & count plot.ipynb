# Heatmap

```python
import pandas as  pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv(r'C:\Users\AFNAN KHAN.DESKTOP-UAK9PNT\Desktop\New folder\penguins.csv')
data
```

```python
# NOTE: heatmap numaric ko support krtahe string ko nhi isliye humne drop comond ke throw string column hta diye he

# Drop the 'species', 'island', and 'sex' columns
data_cleaned = data.drop(columns=['species', 'island', 'sex'])

# Print the resulting DataFrame
print(data_cleaned)
```

```python
sns.heatmap(data_cleaned)   #data_cleaned jo name he wo humne string column name remove krke jo data ka name diya tha wo he yha dataframe ka name change krke kuch or bhi de sakte he
```

```python
sns.heatmap(data_cleaned, vmin=0, vmax=12) 
```

```python
# cmap=  ka use color ke liye tha
# annot=True  --  ye heatmap ke ander number dikhayegha
# linewidth=   line ko choda krne ke liye
#linecolor=

sns.heatmap(data_cleaned, vmin=0, vmax=12,cmap='BrBG', annot=True, )
```
