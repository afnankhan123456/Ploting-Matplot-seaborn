# Strip plot

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
rock = pd.read_csv(r"C:\Users\AFNAN KHAN.DESKTOP-UAK9PNT\Desktop\New folder\tips.csv")
rock
```

```python
sns.stripplot(x='day', y='total_bill', data=rock)
plt.show()
```

```python
sns.stripplot(x='day', y='total_bill', data=rock, hue='sex', palette='rocket_r',edgecolor='r', size=15)
plt.show()
```

```python
sns.stripplot(x='day', y='total_bill', data=rock, hue='sex', palette='rocket_r',edgecolor='r', size=15, marker='*')
plt.show()
```
