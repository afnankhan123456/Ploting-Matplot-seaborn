```python
import seaborn as sns
```

```python
porter = pd.read_csv(r'C:\Users\AFNAN KHAN.DESKTOP-UAK9PNT\Desktop\New folder\tips.csv')
porter
```

```python
#Hue comond se hum day or total bill define krenge time ke throw

import seaborn as sns
import matplotlib.pyplot as plt

sns.violinplot(x="day", y="total_bill", data=porter, hue="time")
plt.show()
```

```python
# linewidth graph ke ander ki line ko bda krna
# palette  comond se bhi color change hota he


sns.violinplot(x="day", y="total_bill", data=porter, hue="time", linewidth=1, palette='RdGy_r')
plt.show()
```

```python
#yha humne time ke base pr lunch or dinner define kiya he

sns.violinplot(x="time", y="total_bill", data=porter,order=['Dinner','lunch'])
plt.show()
```

```python

sns.violinplot(x="day", y="total_bill", data=porter, hue="sex")
plt.show()
```

```python
#split se humne male female ek hi grph me join kr diya


sns.violinplot(x="day", y="total_bill", data=porter, hue="sex",split=True)
plt.show()
```

```python
# inner="quart" comond se hum plot ka design change kr skate he 

sns.violinplot(x="day", y="total_bill", data=porter, hue="sex",split=True, inner="box")
plt.show()
```
