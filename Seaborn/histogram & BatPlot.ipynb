# Barplot

```python
healthcare = pd.read_csv(r'C:\Users\AFNAN KHAN.DESKTOP-UAK9PNT\Desktop\New folder\penguins.csv')
healthcare
```

```python
sns.barplot(x='island', y='bill_depth_mm',data=healthcare)  
plt.show()
```

```python
sns.barplot(x='island', y='bill_depth_mm',data=healthcare, hue='sex')
plt.show()
```

```python
# orient='h' --- is comond se horizontal me change ho jayega

sns.barplot(x='bill_length_mm', y='bill_depth_mm', data=healthcare, hue='sex', orient='h')
plt.show()
```

```python
sns.barplot(x='island', y='bill_depth_mm', data=healthcare, hue='sex', color='b')
plt.show()
```

```python
# palette=  ka use color ke liye hota he 
# saturation= ka use dark or biller ke liye hota he
sns.barplot(x='island', y='bill_depth_mm',data=healthcare, hue='sex', palette='viridis', saturation=0.1,)
plt.show()
```

```python
# errcolor= --- is comon ka use graph ke uper jo err hote he uno color krna hota he
# errwidth= --- is comond se hum error ka waidth bdha sakte he

sns.barplot(x='island', y='bill_depth_mm',data=healthcare, hue='sex', palette='viridis', errcolor='r', errwidth=12)
plt.show()
```
