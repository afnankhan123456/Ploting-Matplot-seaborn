```python
# Q3.write a program that takes a string as input from the user and prints whether the string is a palindrome or not.
# Taking input from the user
string = input("Enter a string: ")

# Removing spaces and converting to lowercase
string = string.lower().replace(" ", "")

# Checking if the string is a palindrome
if string == string[::-1]:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")

```

```python
# Q4. write a program that takes a list of numbers as input from the user and prints the largest number in the list that is divisible by 3.
X = [1, 2, 6, 4, 5]

# Filter the list to include only numbers divisible by 3
y = [num for num in X if num % 3 == 0]

# Check if there are any numbers divisible by 3
if y:
    # Print the largest number divisible by 3
    print("The largest number divisible by 3 is:", max(y))
else:
    print("There are no numbers divisible by 3 in the list.")


```

```python
# 1) Please write a query for Checking the Validity of a Date using Elif

day = int(input("Enter day: "))
month = int(input("Enter month: "))
year = int(input("Enter year: "))

if month in (1, 3, 5, 7, 8, 10, 12):
    max_days = 31
elif month in (4, 6, 9, 11):
    max_days = 30
elif month == 2:
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        max_days = 29
    else:
        max_days = 28
else:
    print("Invalid month")

if 1 <= day <= max_days:
    print("Valid date")
else:
    print("Invalid date")
```

```python
# Q1:  write a query for Determining if a year is a leap year
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

```python
# Q3: write a Python program that takes a list of numbers as input from the user and prints the sum of all the even/odd numbers in list.
x = [1, 2, 4, 5, 3, 7, 8]
sum_even = sum(num for num in x if num % 2 == 0)
sum_odd = sum(num for num in x if num % 2 != 0)
print("Sum of even numbers:", sum_even)
print("Sum of odd numbers:", sum_odd)
```
